```python
while True:
    print("\nMenu:")
    print("1. Perkalian a pangkat b")
    print("2. Hitung deret")
    print("0. Keluar")

    pilihan = int(input("Masukkan pilihan: "))

    if pilihan == 1:
        a = int(input("Masukkan suatu bilangan bulat: "))
        b = int(input("Masukkan pangkat yang diinginkan: "))

        hasil = 1
        for i in range(1, b + 1):
            hasil *= a

        print(f"hasil {a} pangkat {b} adalah {hasil}")

    elif pilihan == 2:
        N = int(input("Masukkan jumlah N: "))

        pembilang = 1
        penyebut = 1
        hasil = 0
        tanda = 1

        for i in range(N):
            hasil += tanda * (pembilang / penyebut)

            temp = pembilang
            pembilang = pembilang + penyebut
            penyebut = temp + pembilang

            tanda *= -1

        print("Hasil =", hasil)

    elif pilihan == 0:
        print("Program selesai")
        break

    else:
        print("Pilihan tidak valid")
```
